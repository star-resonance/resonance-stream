# Chat rules — dedupe, block list, chat log, over the bridge

What the backend does with chat lines, with no game and no clicks: lines go in through `replay-chat`, the chat row's menu and the clear button are
the bridge commands `block-user`, `unblock-user`, `get-chat-history` and `clear-history`, and the result is read from the events the app publishes.
Rows: a repeated line and a second client's copy are shown once; the chat log is what was published; blocking flags the rows already there (and only the
sender's), a later line from the blocked sender, unblocking, clearing. It also asks for the Study view's furigana (`annotate-furigana`): one list of spans per line, the spans joined are the line, a line without kanji stays plain, readings are hiragana. Needs a `test-env` exe and Node 20+; run Jupyter as Administrator. The same pipeline
without Jupyter: `python -m runbook.run chat-rules`.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()              # Jupyter starts a notebook in its own folder:
root = next(p for p in [here, *here.parents] if (p / "runbook" / "common.py").exists())
sys.path.insert(0, str(root))            # ...so find runbook above it
from runbook import bridge, common
from runbook.common import Recorder
from runbook.pipelines.chat_rules import ChatRules
rec = Recorder("chat-rules")
if not common.require_windows_admin(rec):
    raise SystemExit("Start Jupyter from an Administrator terminal, then run again.")
bridge.ensure_installed()                # npm ci in runbook/bridge, the first time
```

```python
EXE = common.ask_exe()          # target/release/resonance-stream.exe of this checkout, unless RUNBOOK_TEXT_LOCAL_EXE says otherwise
ChatRules(rec, EXE, common.RUNS).run()
```
