# Download integrity — the model download refuses what it must

A mock server on this PC stands in for GitHub; the app reads its signed metadata (a throwaway key of this run), and the notebook changes the model
entry it publishes, then plays the setup wizard over the bridge (`download-model`, no arguments: the app takes the url and SHA-256 from the metadata) and watches the `download-result` event, the files on disk and the server's log. A good download installs the model; the model it already has is not
downloaded again; wrong bytes, a cut-off download, a 404, a model with no published SHA-256 and an address that is not https are refused, and after each the installed
model and its folder are exactly as they were. Needs a `test-env` exe and Node 20+; run Jupyter as Administrator. The same pipeline without Jupyter:
`python -m runbook.run download-integrity`.

```python
import sys
from pathlib import Path
here = Path.cwd().resolve()              # Jupyter starts a notebook in its own folder:
root = next(p for p in [here, *here.parents] if (p / "runbook" / "common.py").exists())
sys.path.insert(0, str(root))            # ...so find runbook above it
from runbook import bridge, common
from runbook.common import Recorder
from runbook.pipelines.download_integrity import DownloadIntegrity
rec = Recorder("download-integrity")
if not common.require_windows_admin(rec):
    raise SystemExit("Start Jupyter from an Administrator terminal, then run again.")
bridge.ensure_installed()                # npm ci in runbook/bridge, the first time
```

```python
EXE = common.ask_exe()          # target/release/resonance-stream.exe of this checkout, unless RUNBOOK_TEXT_LOCAL_EXE says otherwise
DownloadIntegrity(rec, EXE, common.RUNS).run()
```
